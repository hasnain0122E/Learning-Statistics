```python
import numpy as np
from scipy import stats
```

```python
n_1 = 30
mean_n1 = 70
n1_std = 8
n_2 = 28
mean_n2 = 74
n2_std = 6
df = min((n_1 -1),(n_2-1))
df
```

```python
t_test = (mean_n1 - mean_n2)/(np.sqrt((np.square(n1_std)/n_1)+(np.square(n2_std)/n_2)))
t_test
```

```python
alpha = 0.05
p_value = 2 * (1 - stats.t.cdf(abs(t_test), df))
p_value
```

```python
if p_value > alpha:
    print("we accept null values")
else:
    print("We reject null values")
```

```python
group_A = [85, 88, 90, 92, 87, 85, 89, 91, 86, 88]
group_B = [82, 84, 80, 83, 81, 79, 78, 85, 84, 83]
mean_groupA = np.mean(group_A)
mean_groupB = np.mean(group_B)
n_groupA = len(group_A)
n_groupB = len(group_B)
groupA_std = np.std(group_A,ddof=1)
groupB_std = np.std(group_B,ddof=1)
df = min((n_groupA - 1),(n_groupB -1))
```

```python
t_test = (mean_groupA - mean_groupB)/(np.sqrt((np.square(groupA_std)/n_groupA)+(np.square(groupB_std)/n_groupB)))
t_test
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_test),df))
p_value
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis (H0)")
else:
    print("Failed to reject the null hypothesis (H1)")
```

```python
t_test1, p_value1 = stats.ttest_ind(group_A, group_B, equal_var= False)
print(t_test1)
print(p_value1)
```

```python
alpha = 0.05
if p_value1 < alpha:
    print("Reject the null hypothesis (H0)")
else:
    print("Failed to reject the null hypothesis (H1)")
```
