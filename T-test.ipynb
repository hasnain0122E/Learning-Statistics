```python
import numpy as np
from scipy import stats
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172] #sample data mostly is <30 but can be greater than 30.
#for t-test the sample std is given not population std
```

```python
population_mean = 170
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)
n = len(sample)
df = n - 1 #df is used to find the p-values to compare with alpha value if p value is > alpha value we accept null hypothesis
```

```python
t_test = (sample_mean - population_mean)/(sample_std/np.sqrt(n))
t_test
```

```python
p_value = 2 * (1- stats.t.cdf(abs(t_test),df))
alpha = 0.05
```

```python
if p_value > alpha:
    print("we accept null hypothesis")
else:
    print("we reject null hypothesis")
```
